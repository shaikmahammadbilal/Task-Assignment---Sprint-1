# Step 1 – Understand the Concept

## 1. Class

A **class** is a blueprint or template used to create objects. It defines the data (attributes) and behavior (methods) that objects can have.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A house blueprint describes how houses should be built. The blueprint is the **class**, while the actual houses are **objects** created from it.

# Step 3 – Implementation Explanation

`Student` is the class. `student1` is an object created from the class. The variables `name` and `age` are attributes of the class.

```python
class Student:
    name = "Bilal"
    age = 22

student1 = Student()

print(student1.name)
print(student1.age)
```

# Step 1 – Understand the Concept

## 2. Object

An **object** is an actual instance of a class. Different objects created from the same class can contain different data.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

If `Student` is a student form, each completed form represents one **object** with its own student information.

# Step 3 – Implementation Explanation

`student1` and `student2` are two separate objects of the `Student` class. Each object stores its own value.

```python
class Student:
    pass

student1 = Student()
student2 = Student()

student1.name = "Bilal"
student2.name = "Rahul"

print(student1.name)
print(student2.name)
```

# Step 1 – Understand the Concept

## 3. Constructor (__init__)

The `__init__()` method is a special method that runs automatically when an object is created. It is commonly used to initialize object data.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

When a new bank account is opened, the account is immediately initialized with an account holder name and account number. `__init__()` performs a similar job.

# Step 3 – Implementation Explanation

`__init__()` receives values when the object is created and stores them in the object using `self`.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Bilal", 22)

print(student1.name)
print(student1.age)
```

# Step 1 – Understand the Concept

## 4. Instance Variables

Instance variables are variables that belong to a particular object. Each object can have different values.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

Every employee has a different name and salary. These values belong to each employee, not to all employees together.

# Step 3 – Implementation Explanation

`self.name` and `self.salary` are instance variables. Their values can be different for every object.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

employee1 = Employee("Bilal", 30000)
employee2 = Employee("Ravi", 40000)

print(employee1.name, employee1.salary)
print(employee2.name, employee2.salary)
```

# Step 1 – Understand the Concept

## 5. Class Variables

A class variable belongs to the class and is shared by all objects of that class.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

If all employees work for the same company, the company name can be shared by every employee.

# Step 3 – Implementation Explanation

`company` is a class variable. Both objects can access the same class-level value.

```python
class Employee:
    company = "ABC Technologies"

    def __init__(self, name):
        self.name = name

employee1 = Employee("Bilal")
employee2 = Employee("Ravi")

print(employee1.company)
print(employee2.company)
```

# Step 1 – Understand the Concept

## 6. Instance Methods

An instance method is a method that works with a particular object. It normally uses `self` as its first parameter.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

An employee can introduce themselves using their own name. The action depends on the individual employee.

# Step 3 – Implementation Explanation

`display()` is an instance method because it uses `self` to access the object's data.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Student Name:", self.name)

student1 = Student("Bilal")
student1.display()
```

# Step 1 – Understand the Concept

## 7. Static Methods

A static method does not need object data or class data. It is defined using `@staticmethod`.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A calculator's addition operation can work with two numbers without knowing anything about a particular calculator object.

# Step 3 – Implementation Explanation

`add()` does not use `self` or `cls`, so it can be a static method.

```python
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))
```

# Step 1 – Understand the Concept

## 8. Class Methods

A class method works with class-level data and uses `cls` as its first parameter. It is defined using `@classmethod`.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

If a company wants to display its common company name, the operation can use information belonging to the class.

# Step 3 – Implementation Explanation

`cls` refers to the class. `show_company()` accesses the class variable through `cls`.

```python
class Employee:
    company = "ABC Technologies"

    @classmethod
    def show_company(cls):
        print("Company:", cls.company)

Employee.show_company()
```

# Step 1 – Understand the Concept

## 9. Properties (@property)

A property allows a method to be accessed like an attribute. It is useful for controlled access to data.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A bank balance may be displayed like a normal value, while the bank controls how that value is calculated or accessed.

# Step 3 – Implementation Explanation

`result` is a property. We access it as `student.result`, without calling it like `student.result()`.

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    @property
    def result(self):
        return "Pass" if self.marks >= 40 else "Fail"

student = Student(75)

print(student.result)
```

# Step 1 – Understand the Concept

## 10. Encapsulation

Encapsulation means keeping data and the methods that operate on that data together, while controlling direct access when needed.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A bank account keeps important account operations inside the banking system instead of allowing anyone to directly change the balance.

# Step 3 – Implementation Explanation

`__balance` uses Python's name-mangling convention for a private-like attribute. The balance is accessed through a method instead of directly.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def show_balance(self):
        print("Balance:", self.__balance)

account = BankAccount(5000)
account.show_balance()
```

# Step 1 – Understand the Concept

## 11. Abstraction

Abstraction means showing only the important details and hiding unnecessary implementation details.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

When you use an ATM, you choose withdraw or deposit. You do not need to know the internal banking software that performs the transaction.

# Step 3 – Implementation Explanation

`Vehicle` defines what `start()` should do, while `Car` provides the actual implementation. The abstract class hides unnecessary details.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car starts with a key")

car = Car()
car.start()
```

# Step 1 – Understand the Concept

## 12. Inheritance

Inheritance allows a child class to reuse attributes and methods from a parent class.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A child can inherit common characteristics from a parent. Similarly, a child class can reuse functionality from a parent class.

# Step 3 – Implementation Explanation

`Dog` inherits `eat()` from `Animal` and also defines its own `bark()` method.

```python
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

dog = Dog()
dog.eat()
dog.bark()
```

# Step 1 – Understand the Concept

## 13. Multiple Inheritance

Multiple inheritance means a child class inherits from more than one parent class.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A smart device may have features from both a camera and a phone.

# Step 3 – Implementation Explanation

`Smartphone` inherits from both `Camera` and `Phone`.

```python
class Camera:
    def take_photo(self):
        print("Taking photo")

class Phone:
    def make_call(self):
        print("Making call")

class Smartphone(Camera, Phone):
    pass

phone = Smartphone()
phone.take_photo()
phone.make_call()
```

# Step 1 – Understand the Concept

## 14. Multilevel Inheritance

Multilevel inheritance is a chain of inheritance where one class inherits from another child class.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

For example: Grandparent → Parent → Child. In software, a class can inherit features through multiple levels.

# Step 3 – Implementation Explanation

`Child` inherits from `Parent`, and `Parent` inherits from `Grandparent`. Therefore, `Child` can access methods from both levels.

```python
class Grandparent:
    def house(self):
        print("Grandparent has a house")

class Parent(Grandparent):
    def car(self):
        print("Parent has a car")

class Child(Parent):
    def bike(self):
        print("Child has a bike")

child = Child()
child.house()
child.car()
child.bike()
```

# Step 1 – Understand the Concept

## 15. Hierarchical Inheritance

Hierarchical inheritance means multiple child classes inherit from the same parent class.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

One common parent type can have multiple specialized children, such as different types of vehicles.

# Step 3 – Implementation Explanation

Both `Dog` and `Cat` inherit the `eat()` method from the same parent class `Animal`.

```python
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

class Cat(Animal):
    def meow(self):
        print("Cat is meowing")

dog = Dog()
cat = Cat()

dog.eat()
cat.eat()
```

# Step 1 – Understand the Concept

## 16. Polymorphism

Polymorphism means the same method or interface can behave differently for different objects.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

The same command, such as `start()`, can start different vehicles in different ways.

# Step 3 – Implementation Explanation

Both objects have a `sound()` method, but each class provides different behavior.

```python
class Dog:
    def sound(self):
        print("Dog says: Woof")

class Cat:
    def sound(self):
        print("Cat says: Meow")

for animal in [Dog(), Cat()]:
    animal.sound()
```

# Step 1 – Understand the Concept

## 17. Method Overriding

Method overriding happens when a child class provides its own implementation of a method already defined in the parent class.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A general vehicle may have a `start()` behavior, while a specific car can provide its own version.

# Step 3 – Implementation Explanation

`Dog` overrides the `sound()` method inherited from `Animal`.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

dog = Dog()
dog.sound()
```

# Step 1 – Understand the Concept

## 18. Method Overloading (Python Approach)

Python does not support traditional method overloading by defining multiple methods with the same name and different parameter lists. Common Python approaches use default arguments or `*args`.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A calculator can accept one, two, or many numbers through one flexible operation.

# Step 3 – Implementation Explanation

One `add()` method accepts a variable number of arguments using `*numbers`. This is a common Python approach to achieve overloading-like behavior.

```python
class Calculator:
    def add(self, *numbers):
        total = 0
        for number in numbers:
            total += number
        return total

calculator = Calculator()

print(calculator.add(10))
print(calculator.add(10, 20))
print(calculator.add(10, 20, 30))
```

# Step 1 – Understand the Concept

## 19. Composition

Composition is a strong 'has-a' relationship where one object contains another object as a part of itself.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A car has an engine. The engine is treated as a component of the car.

# Step 3 – Implementation Explanation

`Car` creates and owns its `Engine` object. This represents a strong has-a relationship.

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()

    def start_car(self):
        self.engine.start()
        print("Car started")

car = Car()
car.start_car()
```

# Step 1 – Understand the Concept

## 20. Aggregation

Aggregation is a weaker 'has-a' relationship where objects can exist independently of the container object.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A department has employees, but an employee can still exist if the department is removed.

# Step 3 – Implementation Explanation

The `Employee` object is created outside `Department` and then passed into it. The employee can exist independently.

```python
class Employee:
    def __init__(self, name):
        self.name = name

class Department:
    def __init__(self, employee):
        self.employee = employee

employee = Employee("Bilal")
department = Department(employee)

print(department.employee.name)
```

# Step 1 – Understand the Concept

## 21. Association

Association means two independent objects are connected or interact with each other, but neither necessarily owns the other.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A teacher teaches a student. Both the teacher and student can exist independently.

# Step 3 – Implementation Explanation

`Teacher` and `Student` are independent objects. They interact through the `teach()` method.

```python
class Teacher:
    def teach(self, student):
        print("Teacher is teaching", student.name)

class Student:
    def __init__(self, name):
        self.name = name

teacher = Teacher()
student = Student("Bilal")

teacher.teach(student)
```

# Step 1 – Understand the Concept

## 22. Magic (Dunder) Methods

Magic methods, also called dunder methods, are special methods whose names start and end with double underscores. Python calls them automatically in certain situations.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

Like special rules that tell an object how to behave when Python performs common operations such as printing, comparing, or getting its length.

# Step 3 – Implementation Explanation

`__str__()` is a dunder method. Python automatically calls it when `print(student)` is used.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student: {self.name}"

student = Student("Bilal")
print(student)
```

# Step 1 – Understand the Concept

## 23. super()

`super()` is used to access methods or the constructor of a parent class from a child class.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A child can use the parent's existing setup and then add its own extra details.

# Step 3 – Implementation Explanation

`super().__init__(name)` calls the parent class constructor, allowing the child class to reuse the parent's initialization.

```python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, course):
        super().__init__(name)
        self.course = course

student = Student("Bilal", "Python")

print(student.name)
print(student.course)
```

# Step 1 – Understand the Concept

## 24. __str__()

`__str__()` defines the user-friendly string representation of an object. It is used when the object is passed to `print()` or `str()`.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A contact card can display a person's name in a readable format instead of showing an internal object address.

# Step 3 – Implementation Explanation

`__str__()` returns a readable string. `print(student)` automatically uses this method.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}, Age: {self.age}"

student = Student("Bilal", 22)
print(student)
```

# Step 1 – Understand the Concept

## 25. __repr__()

`__repr__()` provides a developer-oriented representation of an object. It is useful for debugging and inspecting objects.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

When a developer inspects an object, they usually want a representation that clearly shows what the object contains.

# Step 3 – Implementation Explanation

`repr(student)` calls `__repr__()` and returns a detailed representation useful during development and debugging.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age})"

student = Student("Bilal", 22)
print(repr(student))
```

# Step 1 – Understand the Concept

## 26. __len__()

`__len__()` defines what should be returned when `len()` is called on an object.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A classroom object can define its length as the number of students inside it.

# Step 3 – Implementation Explanation

`len(classroom)` automatically calls `classroom.__len__()` and returns the number of students.

```python
class Classroom:
    def __init__(self, students):
        self.students = students

    def __len__(self):
        return len(self.students)

classroom = Classroom(["Bilal", "Ravi", "Anu"])

print(len(classroom))
```

# Step 1 – Understand the Concept

## 27. Dataclasses (Bonus)

`dataclass` is a convenient Python feature for classes that mainly store data. It can automatically create methods such as `__init__()` and `__repr__()`.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A simple student record can be created without manually writing repetitive constructor and representation code.

# Step 3 – Implementation Explanation

`@dataclass` automatically generates useful methods for the class. Type hints describe the expected data types.

```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
    marks: float

student = Student("Bilal", 22, 85.5)

print(student)
print(student.name)
print(student.marks)
```

# Step 1 – Understand the Concept

## 28. Singleton Pattern (Bonus)

The Singleton Pattern is a design pattern intended to ensure that only one instance of a class is created and reused.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A company may use one shared configuration manager for the entire application instead of creating many separate configuration managers.

# Step 3 – Implementation Explanation

`__new__()` controls object creation. The first object is stored in `_instance`; later calls return the same object. Therefore, `object1 is object2` is `True`.

```python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

object1 = Singleton()
object2 = Singleton()

print(object1 is object2)
```
