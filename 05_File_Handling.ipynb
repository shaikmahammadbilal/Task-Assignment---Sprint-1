## 1. Reading Text Files

### Definition
Reading a text file means opening an existing file and getting its contents into a Python program.

### Common Methods
- `open()` – opens a file
- `read()` – reads the complete file
- `readline()` – reads one line
- `readlines()` – reads all lines as a list

### Real-Time Example
Reading a student's notes from a text file.


````python
# Create a sample text file first
with open("student_notes.txt", "w") as file:
    file.write("Python File Handling\n")
    file.write("Python is easy to learn.\n")
    file.write("Practice every day.\n")

# Read the complete file
file = open("student_notes.txt", "r")
content = file.read()
print(content)
file.close()
````

````python
# Reading one line at a time
file = open("student_notes.txt", "r")

print("First line:")
print(file.readline())

print("Second line:")
print(file.readline())

file.close()
````

````python
# Reading all lines into a list
file = open("student_notes.txt", "r")
lines = file.readlines()
file.close()

print(lines)
````

## 2. Writing Text Files

### Definition
Writing means storing text or other string data inside a file.

### File Modes
- `"w"` – write; creates a new file or replaces existing content
- `"a"` – append; adds content at the end
- `"x"` – creates a new file and gives an error if it already exists

### Real-Time Example
Saving a user's registration information into a text file.


````python
# Writing to a file
with open("user.txt", "w") as file:
    file.write("Name: Bilal\n")
    file.write("Role: Software Engineer\n")

print("Data written successfully.")
````

````python
# Appending data
with open("user.txt", "a") as file:
    file.write("Skill: Python\n")

print("New data added successfully.")

# Read the updated file
with open("user.txt", "r") as file:
    print(file.read())
````

## 3. CSV Files

### Definition
CSV stands for **Comma-Separated Values**. It is commonly used to store tabular data such as employee, student, or sales information.

Python provides the built-in `csv` module.

### Real-Time Example
Storing student details in rows and columns.


````python
import csv

# Writing CSV data
students = [
    ["ID", "Name", "Marks"],
    [1, "Bilal", 85],
    [2, "Rahul", 90],
    [3, "Anu", 78]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(students)

print("CSV file created successfully.")
````

````python
# Reading CSV data
with open("students.csv", "r", newline="") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
````

````python
# Writing CSV using dictionaries
student_data = [
    {"id": 1, "name": "Bilal", "marks": 85},
    {"id": 2, "name": "Rahul", "marks": 90}
]

with open("student_details.csv", "w", newline="") as file:
    fieldnames = ["id", "name", "marks"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(student_data)

print("Dictionary data saved to CSV.")
````

## 4. JSON Files

### Definition
JSON stands for **JavaScript Object Notation**. It is a common format for storing and exchanging structured data.

Python provides the built-in `json` module.

### Common Functions
- `json.dump()` – writes Python data to a JSON file
- `json.load()` – reads JSON data from a file
- `json.dumps()` – converts Python data to a JSON string
- `json.loads()` – converts a JSON string to Python data

### Real-Time Example
Saving user profile information or API data.


````python
import json

user = {
    "name": "Bilal",
    "role": "Software Engineer",
    "skills": ["Python", "SQL", "React"]
}

# Write Python dictionary to JSON file
with open("user.json", "w") as file:
    json.dump(user, file, indent=4)

print("JSON file created successfully.")
````

````python
# Read JSON file
with open("user.json", "r") as file:
    data = json.load(file)

print(data)
print("Name:", data["name"])
print("Skills:", data["skills"])
````

````python
# JSON string example
json_text = '{"name": "Bilal", "age": 25}'

data = json.loads(json_text)

print(data)
print(data["name"])
````

## 5. Pickle Files

### Definition
`pickle` is a Python module used to serialize and deserialize Python objects.

- **Serialization**: Python object → file
- **Deserialization**: file → Python object

Pickle can store Python-specific objects such as lists, dictionaries, and custom objects.

### Important Safety Note
Never unpickle data from an untrusted or unknown source because pickle can execute malicious code.

### Real-Time Example
Saving Python program data so it can be loaded later.


````python
import pickle

employee = {
    "id": 101,
    "name": "Bilal",
    "skills": ["Python", "SQL"]
}

# Save object to a pickle file
with open("employee.pkl", "wb") as file:
    pickle.dump(employee, file)

print("Object saved successfully.")
````

````python
# Load object from pickle file
with open("employee.pkl", "rb") as file:
    loaded_employee = pickle.load(file)

print(loaded_employee)
print("Employee name:", loaded_employee["name"])
````

## 6. Context Manager (`with`)

### Definition
A context manager automatically manages resources such as files.

The `with` statement automatically closes the file after the block finishes, even when an exception occurs.

### Why Use It?
Without `with`, we need to manually call `file.close()`.

With `with`, Python handles closing the file for us.


````python
# Recommended way to work with files
with open("example.txt", "w") as file:
    file.write("Hello from Python!")

# The file is automatically closed here
print("File operation completed.")
````

````python
# Checking whether the file is closed
with open("example.txt", "r") as file:
    print("Inside with block:", file.closed)

print("Outside with block:", file.closed)
````

## 7. `pathlib`

### Definition
`pathlib` provides an object-oriented way to work with files and directories.

It is modern, readable, and works across operating systems.

### Common Methods
- `Path.cwd()` – current directory
- `Path.exists()` – checks existence
- `Path.is_file()` – checks whether it is a file
- `Path.is_dir()` – checks whether it is a directory
- `Path.mkdir()` – creates a directory
- `Path.iterdir()` – lists directory contents
- `Path.read_text()` – reads text
- `Path.write_text()` – writes text


````python
from pathlib import Path

# Current working directory
current_path = Path.cwd()
print("Current directory:", current_path)

# Create and write a file
file_path = Path("pathlib_example.txt")
file_path.write_text("Hello using pathlib!")

print("File exists:", file_path.exists())
print("Is file:", file_path.is_file())

# Read the file
print(file_path.read_text())
````

````python
# Create a directory using pathlib
folder = Path("pathlib_folder")
folder.mkdir(exist_ok=True)

print("Directory created:", folder)
print("Is directory:", folder.is_dir())
````

## 8. `os`

### Definition
The `os` module provides functions for interacting with the operating system.

### Common Uses
- Get current directory
- List files and folders
- Create directories
- Rename files
- Remove files
- Work with environment variables


````python
import os

# Current working directory
print("Current directory:", os.getcwd())

# List files and folders
print("Directory contents:")
print(os.listdir())
````

````python
# Create a directory using os
os.makedirs("os_example_folder", exist_ok=True)

print("Directory created successfully.")

# Check whether a path exists
print(os.path.exists("os_example_folder"))

# Check whether it is a directory
print(os.path.isdir("os_example_folder"))
````

````python
# Create and rename a file
with open("old_name.txt", "w") as file:
    file.write("Sample data")

os.rename("old_name.txt", "new_name.txt")

print("File renamed successfully.")
````

## 9. `shutil`

### Definition
`shutil` stands for **shell utilities**. It provides high-level operations for copying, moving, and deleting files and directories.

### Common Functions
- `shutil.copy()` – copy a file
- `shutil.copytree()` – copy a directory
- `shutil.move()` – move a file or directory
- `shutil.rmtree()` – remove a directory and its contents

### Real-Time Example
Creating a backup copy of an important file.


````python
import shutil

# Create a source file
with open("source.txt", "w") as file:
    file.write("Important project data")

# Copy the file
shutil.copy("source.txt", "backup_source.txt")

print("File copied successfully.")
````

````python
# Move a file
shutil.move("backup_source.txt", "moved_backup.txt")

print("File moved successfully.")
````

## 10. File Validation

### Definition
File validation means checking whether a file exists and whether it has the expected properties before using it.

### Common Checks
- Does the file exist?
- Is it actually a file?
- Is it empty?
- Does it have the correct extension?
- Is the file readable?

### Real-Time Example
Before processing an uploaded CSV file, a program can validate its existence and extension.


````python
from pathlib import Path

file_path = Path("students.csv")

if not file_path.exists():
    print("File does not exist.")
elif not file_path.is_file():
    print("Path is not a file.")
elif file_path.stat().st_size == 0:
    print("File is empty.")
else:
    print("File exists and contains data.")
````

````python
# Validate file extension
file_path = Path("students.csv")

if file_path.suffix.lower() == ".csv":
    print("Valid CSV file.")
else:
    print("Invalid file type.")
````

````python
# Check whether a file can be read
file_path = Path("students.csv")

if file_path.is_file():
    try:
        with open(file_path, "r") as file:
            file.read(1)
        print("File is readable.")
    except PermissionError:
        print("Permission denied.")
else:
    print("File not found.")
````

## 11. Directory Operations

### Definition
Directory operations are actions performed on folders, such as creating, listing, renaming, moving, and deleting directories.

### Common Operations
- Create a directory
- Check whether a directory exists
- List directory contents
- Rename a directory
- Remove a directory
- Traverse folders

### Real-Time Example
A project may create separate folders for reports, images, and backups.


````python
from pathlib import Path

# Create directories
project_folder = Path("project_files")
reports_folder = project_folder / "reports"

project_folder.mkdir(exist_ok=True)
reports_folder.mkdir(exist_ok=True)

print("Project directories created.")
````

````python
# List directory contents
project_folder = Path("project_files")

print("Contents of project_files:")

for item in project_folder.iterdir():
    print(item)
````

````python
# Create a file inside the reports folder
report_file = Path("project_files") / "reports" / "report.txt"
report_file.write_text("Monthly project report")

print("Report file created.")
````

````python
# Rename a directory
old_folder = Path("project_files")
new_folder = Path("my_project_files")

if old_folder.exists() and not new_folder.exists():
    old_folder.rename(new_folder)
    print("Directory renamed successfully.")
else:
    print("Rename skipped because the source does not exist or destination already exists.")
````

| `t` | Text mode |
| `r+` | Read and write |
