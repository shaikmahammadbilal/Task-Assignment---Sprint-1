# 1. Iterators

An iterator is an object that allows us to access elements one by one.

Python provides two important functions:

- `iter()` → creates an iterator
- `next()` → gets the next value

An iterator remembers its current position.

### Real-Time Example

Suppose an application has a list of customer names.

Instead of accessing all customers at once, an iterator can provide
one customer at a time.

```python
customers = ["Bilal", "Rahul", "John"]

customer_iterator = iter(customers)

print(next(customer_iterator))
print(next(customer_iterator))
print(next(customer_iterator))
```

# 2. Generators

A generator is a special type of function that produces values one at a time.

Generators are useful when working with large amounts of data because
they do not store all values in memory at the same time.

### Real-Time Example

Imagine a company has 1 million customer records.

If we create a normal list containing all 1 million records,
the application may use a lot of memory.

A generator can provide one record at a time.

```python
def generate_numbers():
    for number in range(1, 6):
        yield number


for number in generate_numbers():
    print(number)
```

# 3. yield

The `yield` keyword is used inside a generator.

Unlike `return`, `yield` does not completely stop the function.

It pauses the function and remembers its current position.

When the generator is called again, it continues from where it stopped.

### Difference

return:
- Returns a value
- Function ends

yield:
- Produces a value
- Function pauses
- Can continue later

```python
def simple_generator():
    yield 10
    yield 20
    yield 30


numbers = simple_generator()

print(next(numbers))
print(next(numbers))
print(next(numbers))
```

# 6. global

A global variable is created outside a function.

Normally, a function can read a global variable.

The `global` keyword is used when we want to modify a global
variable from inside a function.

### Real-Time Example

Imagine an application maintaining a global counter for the
number of users who have logged in.

```python
login_count = 0


def user_login():
    global login_count

    login_count += 1


user_login()
user_login()
user_login()

print("Total Logins:", login_count)
```

# 7. nonlocal

The `nonlocal` keyword is used inside a nested function.

It allows the inner function to modify a variable belonging
to the outer function.

### Real-Time Example

A counter function can remember and update its count every time
the returned function is called.

```python
def create_counter():

    count = 0

    def counter():
        nonlocal count

        count += 1
        return count

    return counter


counter = create_counter()

print(counter())
print(counter())
print(counter())
```

## 1. map()

### What is map()?

The `map()` function applies a function to every item in a collection.

### Syntax:

map(function, iterable)

It returns a map object, which can be converted into a list.

### Real-time use:

`map()` can be used when we need to perform the same operation on many values.

Example:
- Add tax to product prices
- Convert temperatures
- Calculate salaries
- Modify customer data

```python
numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x * 2, numbers)

print(list(result))
```

## 2. filter()

### What is filter()?

The `filter()` function selects items from a collection based on a condition.

Only items for which the condition is `True` are returned.

### Syntax:

filter(function, iterable)

### Real-time use:

`filter()` can be used to find:
- Employees with salary above a certain amount
- Students who passed
- Products below a price
- Active users

```python
numbers = [10, 15, 20, 25, 30]

result = filter(lambda x: x > 20, numbers)

print(list(result))
```

## 3. zip()

### What is zip()?

The `zip()` function combines two or more collections together item by item.

### Real-time use:

`zip()` is useful when we have related data in separate lists.

Example:
- Employee names + salaries
- Student names + marks
- Product names + prices

```python
names = ["Bilal", "Rahul", "John"]
marks = [85, 90, 75]

result = zip(names, marks)

print(list(result))
```

## 6. any()

### What is any()?

The `any()` function returns `True` if at least one value in a collection is True.

If all values are False, it returns False.

### Real-time use:

It can be used to check:
- Is at least one student passed?
- Is at least one product available?
- Is at least one employee active?

```python
numbers = [10, 20, 30, 40]

result = any(number > 35 for number in numbers)

print(result)
```

## 7. all()

### What is all()?

The `all()` function returns `True` only when all values are True.

If even one value is False, it returns False.

### Real-time use:

It can be used to check:
- Did all students pass?
- Are all products available?
- Are all required fields filled?

```python
marks = [60, 70, 80, 90]

result = all(mark >= 50 for mark in marks)

print(result)
```
