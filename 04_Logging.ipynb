# Step 1 – Understand the Concept

## 1. What is Logging?

Logging is the process of recording information about what a program is doing while it runs. Python provides the built-in `logging` module.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A security camera records important events in a building. Logging records important events happening inside an application.

# Step 3 – Implementation Explanation

`logging` can record application startup, normal operations, warnings, errors, and critical failures.

````python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started")
logging.warning("This is a warning")
logging.error("An error occurred")
````

# Step 1 – Understand the Concept

## 2. Why Use Logging?

Logging helps developers monitor applications, troubleshoot problems, understand program behavior, and keep a history of important events.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A hospital keeps records of important events so staff can understand what happened later.

# Step 3 – Implementation Explanation

Logging is useful because developers do not need to watch the application continuously.

````python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("User logged in")
logging.info("Order created successfully")
logging.warning("Invalid value entered")
logging.error("Database connection failed")
````

# Step 1 – Understand the Concept

## 3. `print()` vs `logging`

`print()` displays simple output. `logging` provides levels, timestamps, handlers, formatting, and multiple destinations.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

`print()` is like speaking a message aloud. Logging is like maintaining an organized official record.

# Step 3 – Implementation Explanation

Use `print()` for simple temporary output. Use `logging` for application monitoring and debugging.

````python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s: %(message)s"
)

print("This is a print message")
logging.info("This is a logging message")
logging.warning("This is a warning")
logging.error("This is an error")
````

## 4. Log Levels

Log levels show the importance or severity of a log message.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

Traffic signals communicate different levels of attention. Log levels do something similar for software events.

# Step 3 – Implementation Explanation

The common levels are `DEBUG`, `INFO`, `WARNING`, `ERROR`, and `CRITICAL`.

````python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(levelname)s: %(message)s"
)

logging.debug("Debug message")
logging.info("Information message")
logging.warning("Warning message")
logging.error("Error message")
logging.critical("Critical message")
````

# Step 1 – Understand the Concept

## 10. Log Formatters

A formatter controls how log messages are displayed or stored. It can include time, level, logger name, and message.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A receipt has a fixed structure such as date, item, and price. A formatter gives logs a consistent structure.

# Step 3 – Implementation Explanation

`logging.Formatter()` defines the structure of each log record.

````python
import logging

logger = logging.getLogger("MyApp")
logger.setLevel(logging.DEBUG)

handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(name)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

logger.info("Application started")
logger.warning("Low memory warning")
````

# Step 1 – Understand the Concept

## 11. File Handlers

A `FileHandler` sends log messages to a file.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A company may write important events into a record book so they can be reviewed later.

# Step 3 – Implementation Explanation

`logging.FileHandler()` writes log records to a file.

````python
import logging

logger = logging.getLogger("FileLogger")
logger.setLevel(logging.INFO)

handler = logging.FileHandler("application.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

logger.info("This message is stored in application.log")
logger.error("An error was recorded")
````

# Step 1 – Understand the Concept

## 12. Console Handlers

A console handler sends log messages to the console or another stream.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A manager can announce an event directly to people instead of only writing it in a record book.

# Step 3 – Implementation Explanation

`logging.StreamHandler()` is commonly used for terminal or console output.

````python
import logging

logger = logging.getLogger("ConsoleLogger")
logger.setLevel(logging.DEBUG)

handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(levelname)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

logger.debug("Debug information")
logger.info("Application running")
logger.warning("Warning message")
````

# Step 1 – Understand the Concept

## 13. File and Console Handlers Together

A logger can send records to multiple destinations, such as both a file and the console.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A company can announce an event to employees and also record it in an official register.

# Step 3 – Implementation Explanation

Create separate handlers and attach both to the same logger.

````python
import logging

logger = logging.getLogger("Application")
logger.setLevel(logging.DEBUG)

console_handler = logging.StreamHandler()
file_handler = logging.FileHandler("application.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)
file_handler.setFormatter(formatter)

logger.addHandler(console_handler)
logger.addHandler(file_handler)

logger.info("Message goes to console and file")
logger.error("Error goes to console and file")
````

# Step 1 – Understand the Concept

## 14. Rotating Logs (Bonus)

Rotating logs means creating backup log files when the current file reaches a configured size or time limit.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A shop replaces a full register book with a new book while keeping the old books for records.

# Step 3 – Implementation Explanation

`RotatingFileHandler` can automatically create backup files when the current log reaches a size limit.

````python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("RotatingLogger")
logger.setLevel(logging.INFO)

handler = RotatingFileHandler(
    "rotating_application.log",
    maxBytes=1000,
    backupCount=3
)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

for number in range(100):
    logger.info("Log message number %s", number)

print("Rotating log example completed.")
````

## 15. Logging Best Practices

Logging best practices make logs useful, readable, secure, and manageable.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A well-organized office record uses consistent labels, stores useful information, and protects confidential information.

# Step 3 – Implementation Explanation

Use appropriate levels, meaningful messages, timestamps, formatters, and rotating logs when needed. Never log passwords, API keys, tokens, or other sensitive data.

````python
import logging

logger = logging.getLogger("BestPracticeApp")
logger.setLevel(logging.INFO)

handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(name)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

username = "Bilal"

logger.info("User login successful: %s", username)
logger.warning("Login attempt failed")
logger.error("Unable to connect to database")

# Never log passwords, API keys, tokens, or other sensitive information.
````
