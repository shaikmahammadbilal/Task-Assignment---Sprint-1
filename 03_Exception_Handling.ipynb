# Step 1 – Understand the Concept

## 1. try

`try` contains code that may cause an exception.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

It is like identifying a risky step in a process before doing it.

# Step 3 – Implementation Explanation

Python executes the code inside `try`. If an exception occurs, Python looks for a matching `except` block.

````python
try:
    number = int("100")
    print("Number:", number)
except ValueError:
    print("Invalid number")
````

# Step 1 – Understand the Concept

## 2. except

`except` handles an exception raised inside a `try` block.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

If something goes wrong, an emergency procedure tells you what to do next.

# Step 3 – Implementation Explanation

The `except` block prevents the program from stopping unexpectedly and lets us show a useful message.

````python
try:
    number = int("abc")
except ValueError:
    print("Please enter a valid number.")
````

# Step 1 – Understand the Concept

## 3. else

`else` runs only when no exception occurs in the `try` block.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

When a task succeeds, you can continue with the next step.

# Step 3 – Implementation Explanation

Use `else` for code that should execute only after the risky operation succeeds.

````python
try:
    number = int("50")
except ValueError:
    print("Invalid input")
else:
    print("Conversion successful")
    print("Number:", number)
````

# Step 1 – Understand the Concept

## 4. finally

`finally` always runs, whether an exception occurs or not.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

After using a machine, cleanup may be required whether the job succeeds or fails.

# Step 3 – Implementation Explanation

`finally` is commonly used for cleanup such as closing files or releasing resources.

````python
try:
    number = int("abc")
except ValueError:
    print("Invalid input")
finally:
    print("This block always runs.")
````

# Step 1 – Understand the Concept

## 5. raise

`raise` manually creates an exception when a program rule is violated.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A security guard can stop an action when a rule is broken.

# Step 3 – Implementation Explanation

Here, a negative age is invalid, so the program raises `ValueError`.

````python
age = -5

try:
    if age < 0:
        raise ValueError("Age cannot be negative")
    print("Age:", age)
except ValueError as error:
    print("Error:", error)
````

# Step 1 – Understand the Concept

## 6. assert

`assert` checks whether a condition is true. If false, Python raises `AssertionError`.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

Before submitting a form, you may check that an important value is valid.

# Step 3 – Implementation Explanation

`assert condition, message` is useful for checking assumptions during development.

````python
marks = 75

try:
    assert 0 <= marks <= 100, "Marks must be between 0 and 100"
    print("Valid marks:", marks)
except AssertionError as error:
    print("Assertion failed:", error)
````

# Step 1 – Understand the Concept

## 7. Custom Exceptions

A custom exception is a user-defined exception class for an application-specific error.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

A bank may use a specific error such as 'Insufficient Balance' instead of a generic error.

# Step 3 – Implementation Explanation

Custom exceptions normally inherit from `Exception` and make application errors easier to understand.

````python
class InsufficientBalanceError(Exception):
    pass

balance = 1000
withdraw_amount = 1500

try:
    if withdraw_amount > balance:
        raise InsufficientBalanceError("Insufficient balance")
    balance -= withdraw_amount
    print("Remaining balance:", balance)
except InsufficientBalanceError as error:
    print("Error:", error)
````

### When it occurs

# Step 1 – Understand the Concept

## Built-in Exception – AttributeError

`AttributeError` occurs when an object does not have the requested attribute or method.

# Step 2 – Demonstrate the Concept

### Real-World Analogy

Trying to use a feature that a device does not provide is similar.

# Step 3 – Implementation Explanation

### When it occurs

`AttributeError` occurs when an object does not have the requested attribute or method.

### Correct Handling

The `Student` object has `name`, but no `salary` attribute.

````python
class Student:
    def __init__(self, name):
        self.name = name

student = Student("Bilal")

try:
    print(student.salary)
except AttributeError:
    print("AttributeError handled: No 'salary' attribute.")
````

# Step 2 – Demonstrate the Concept

# Step 3 – Implementation Explanation

`FileNotFoundError` occurs when a requested file does not exist.

# Bonus – Multiple Exceptions

Different exceptions can be handled separately using multiple `except` blocks.

````python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print("Result:", result)

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Calculation completed successfully.")

finally:
    print("Program execution completed.")
````

# Exception Handling Quick Revision

| Topic | Main Idea |
|---|---|
| `try` | Code that may cause an exception |
| `except` | Handles an exception |
| `else` | Runs when no exception occurs |
| `finally` | Always runs |
| `raise` | Manually raises an exception |
| `assert` | Checks a condition |
| Custom Exception | User-defined application error |
| `ValueError` | Invalid value |
| `TypeError` | Wrong/incompatible data type |
| `IndexError` | Invalid sequence index |
| `KeyError` | Missing dictionary key |
| `AttributeError` | Missing object attribute/method |
| `NameError` | Name not defined |
| `FileNotFoundError` | File does not exist |
| `ZeroDivisionError` | Division by zero |
| `ImportError` | Import request cannot be satisfied |
| `ModuleNotFoundError` | Module cannot be found |

## Basic Pattern

```python
try:
    # risky code
    pass
except SomeException:
    # handle error
    pass
else:
    # success path
    pass
finally:
    # always runs
    pass
```
